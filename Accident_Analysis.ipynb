```python
# ============================================================
# STEP 1: IMPORT PACKAGES
# RTA / ACCIDENT DATASET ANALYSIS
# ============================================================

# ------------------------------------------------------------
# DATA MANIPULATION
# ------------------------------------------------------------

import pandas as pd
import numpy as np


# ------------------------------------------------------------
# DATA VISUALIZATION
# ------------------------------------------------------------

import matplotlib.pyplot as plt
import seaborn as sns


# ------------------------------------------------------------
# STATISTICAL ANALYSIS
# ------------------------------------------------------------

from scipy import stats


# ------------------------------------------------------------
# DISPLAY SETTINGS
# ------------------------------------------------------------

# Display all columns when viewing DataFrames
pd.set_option('display.max_columns', None)

# Display up to 100 rows when required
pd.set_option('display.max_rows', 100)

# Display numerical values up to 3 decimal places
pd.set_option('display.float_format', lambda x: '%.3f' % x)


# ------------------------------------------------------------
# VISUALIZATION SETTINGS
# ------------------------------------------------------------

# Set professional visualization style
sns.set_theme(style="whitegrid")

# Default figure size
plt.rcParams['figure.figsize'] = (10, 6)

# Improve chart readability
plt.rcParams['axes.titlesize'] = 14
plt.rcParams['axes.labelsize'] = 11
plt.rcParams['xtick.labelsize'] = 10
plt.rcParams['ytick.labelsize'] = 10


# ------------------------------------------------------------
# CONFIRMATION
# ------------------------------------------------------------

print("All required packages have been imported successfully.")
```

```python
# ============================================================
# STEP 2: READ DATA & INITIAL DATA UNDERSTANDING
# RTA / ACCIDENT DATASET ANALYSIS
# ============================================================


# ------------------------------------------------------------
# 2.1 IMPORT / UPLOAD DATASET
# ------------------------------------------------------------
# ------------------------------------------------------------
# 2.2 READ THE CSV FILE
# ------------------------------------------------------------

file_name = "RTA_Dataset.csv"

df = pd.read_csv("/content/RTA_Dataset.csv")


# ------------------------------------------------------------
# 2.3 CREATE A BACKUP OF THE ORIGINAL DATA
# ------------------------------------------------------------

# Keeping a copy allows us to compare the cleaned dataset
# with the original dataset later.

df_original = df.copy()


# ------------------------------------------------------------
# 2.4 DISPLAY BASIC INFORMATION
# ------------------------------------------------------------

print("=" * 70)
print("DATASET LOADED SUCCESSFULLY")
print("=" * 70)

print(f"\nDataset Name      : {file_name}")
print(f"Number of Rows    : {df.shape[0]:,}")
print(f"Number of Columns : {df.shape[1]:,}")


# ------------------------------------------------------------
# 2.5 DISPLAY FIRST 5 RECORDS
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("FIRST 5 RECORDS")
print("=" * 70)

display(df.head())


# ------------------------------------------------------------
# 2.6 DISPLAY LAST 5 RECORDS
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("LAST 5 RECORDS")
print("=" * 70)

display(df.tail())


# ------------------------------------------------------------
# 2.7 DATASET DIMENSIONS
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("DATASET DIMENSIONS")
print("=" * 70)

print(f"Rows    : {df.shape[0]:,}")
print(f"Columns : {df.shape[1]:,}")


# ------------------------------------------------------------
# 2.8 COLUMN NAMES
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("COLUMN NAMES")
print("=" * 70)

for i, column in enumerate(df.columns, start=1):
    print(f"{i:2}. {column}")


# ------------------------------------------------------------
# 2.9 DATA TYPES
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("DATA TYPES")
print("=" * 70)

display(
    pd.DataFrame({
        "Column": df.columns,
        "Data_Type": df.dtypes.astype(str).values
    })
)


# ------------------------------------------------------------
# 2.10 DATASET INFORMATION
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("DATASET INFORMATION")
print("=" * 70)

df.info()


# ------------------------------------------------------------
# 2.11 NUMERICAL COLUMNS
# ------------------------------------------------------------

numerical_columns = df.select_dtypes(
    include=np.number
).columns.tolist()

print("\n" + "=" * 70)
print("NUMERICAL COLUMNS")
print("=" * 70)

print(f"Number of numerical columns: {len(numerical_columns)}")

for column in numerical_columns:
    print(f"- {column}")


# ------------------------------------------------------------
# 2.12 CATEGORICAL / OBJECT COLUMNS
# ------------------------------------------------------------

categorical_columns = df.select_dtypes(
    include=["object", "category"]
).columns.tolist()

print("\n" + "=" * 70)
print("CATEGORICAL / OBJECT COLUMNS")
print("=" * 70)

print(f"Number of categorical columns: {len(categorical_columns)}")

for column in categorical_columns:
    print(f"- {column}")


# ------------------------------------------------------------
# 2.13 STATISTICAL SUMMARY - NUMERICAL VARIABLES
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("STATISTICAL SUMMARY - NUMERICAL VARIABLES")
print("=" * 70)

if len(numerical_columns) > 0:
    display(
        df[numerical_columns].describe().T
    )
else:
    print("No numerical columns detected.")


# ------------------------------------------------------------
# 2.14 STATISTICAL SUMMARY - CATEGORICAL VARIABLES
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("STATISTICAL SUMMARY - CATEGORICAL VARIABLES")
print("=" * 70)

if len(categorical_columns) > 0:
    display(
        df[categorical_columns].describe().T
    )
else:
    print("No categorical columns detected.")


# ------------------------------------------------------------
# 2.15 UNIQUE VALUES IN EACH COLUMN
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("NUMBER OF UNIQUE VALUES PER COLUMN")
print("=" * 70)

unique_summary = pd.DataFrame({
    "Column": df.columns,
    "Unique_Values": df.nunique(dropna=False).values,
    "Missing_Values": df.isna().sum().values
})

display(
    unique_summary.sort_values(
        by="Unique_Values",
        ascending=False
    ).reset_index(drop=True)
)


# ------------------------------------------------------------
# 2.16 SAMPLE OF UNIQUE VALUES
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("SAMPLE UNIQUE VALUES FOR EACH COLUMN")
print("=" * 70)

for column in df.columns:

    unique_values = df[column].dropna().unique()

    print(f"\n{column}")
    print("-" * len(column))

    # Show maximum 15 unique values to avoid excessive output
    print(unique_values[:15])

    if len(unique_values) > 15:
        print(f"... and {len(unique_values) - 15} more unique values")


# ------------------------------------------------------------
# 2.17 MISSING VALUE OVERVIEW
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("MISSING VALUE OVERVIEW")
print("=" * 70)

missing_summary = pd.DataFrame({
    "Column": df.columns,
    "Missing_Count": df.isnull().sum().values,
    "Missing_Percentage": (
        df.isnull().mean().values * 100
    )
})

missing_summary = missing_summary.sort_values(
    by="Missing_Count",
    ascending=False
).reset_index(drop=True)

display(missing_summary)


# ------------------------------------------------------------
# 2.18 DUPLICATE RECORD COUNT
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("DUPLICATE RECORD CHECK")
print("=" * 70)

duplicate_count = df.duplicated().sum()

print(f"Number of duplicate rows: {duplicate_count:,}")

if duplicate_count == 0:
    print("No exact duplicate records detected.")
else:
    print("Duplicate records detected and will be investigated in Data Cleaning.")


# ------------------------------------------------------------
# 2.19 TARGET VARIABLE INITIAL CHECK
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("ACCIDENT SEVERITY - INITIAL DISTRIBUTION")
print("=" * 70)

if "Accident_severity" in df.columns:

    severity_summary = pd.DataFrame({
        "Count": df["Accident_severity"].value_counts(dropna=False),
        "Percentage": (
            df["Accident_severity"]
            .value_counts(normalize=True, dropna=False) * 100
        )
    })

    display(severity_summary)

else:
    print("Accident_severity column not found.")


# ------------------------------------------------------------
# 2.20 CASUALTY SUMMARY
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("NUMBER OF CASUALTIES - INITIAL SUMMARY")
print("=" * 70)

if "Number_of_casualties" in df.columns:

    display(
        df["Number_of_casualties"].describe()
    )

    print(
        f"\nTotal casualties recorded: "
        f"{df['Number_of_casualties'].sum():,.0f}"
    )

else:
    print("Number_of_casualties column not found.")


# ------------------------------------------------------------
# 2.21 TIME COLUMN INITIAL INSPECTION
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("TIME COLUMN - INITIAL INSPECTION")
print("=" * 70)

if "Time" in df.columns:

    print(f"Data type of Time column: {df['Time'].dtype}")

    print("\nSample time values:")
    display(df["Time"].head(10))

    print("\nNumber of unique time values:")
    print(df["Time"].nunique())

else:
    print("Time column not found.")


# ------------------------------------------------------------
# 2.22 DATASET MEMORY USAGE
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("DATASET MEMORY USAGE")
print("=" * 70)

memory_usage = df.memory_usage(deep=True).sum() / (1024 ** 2)

print(f"Memory used by dataset: {memory_usage:.2f} MB")


# ------------------------------------------------------------
# 2.23 FINAL INITIAL CHECK
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("STEP 2 COMPLETED")
print("=" * 70)

print(f"Dataset shape : {df.shape}")
print(f"Rows          : {len(df):,}")
print(f"Columns       : {len(df.columns)}")
print(f"Duplicates    : {duplicate_count:,}")
print(f"Memory usage  : {memory_usage:.2f} MB")

print("\nThe dataset is now loaded and initially understood.")
print("Detailed cleaning and treatment will be performed in Step 3.")
```

```python
# ============================================================
# STEP 3: DATA CLEANING
# RTA / ACCIDENT DATASET ANALYSIS
# ============================================================


# ============================================================
# 3.1 CREATE A WORKING COPY
# ============================================================

# Keep df_original untouched.
# All cleaning operations will be performed on df.

df = df_original.copy()

print("=" * 70)
print("DATA CLEANING STARTED")
print("=" * 70)


# ============================================================
# 3.2 CHECK DATASET SHAPE BEFORE CLEANING
# ============================================================

print("\n" + "=" * 70)
print("DATASET SHAPE BEFORE CLEANING")
print("=" * 70)

print(f"Rows    : {df.shape[0]:,}")
print(f"Columns : {df.shape[1]}")
```

```python
# ============================================================
# 3.3 STANDARDIZE COLUMN NAMES
# ============================================================

# Remove accidental spaces from column names
# and replace spaces with underscores.

df.columns = (
    df.columns
    .str.strip()
    .str.replace(" ", "_")
)

print("\n" + "=" * 70)
print("COLUMN NAMES AFTER STANDARDIZATION")
print("=" * 70)

print(df.columns.tolist())
```

```python
# ============================================================
# 3.4 CHECK MISSING VALUES
# ============================================================

print("\n" + "=" * 70)
print("MISSING VALUES BEFORE TREATMENT")
print("=" * 70)

missing_before = pd.DataFrame({
    "Column": df.columns,
    "Missing_Count": df.isnull().sum().values,
    "Missing_Percentage": (
        df.isnull().mean().values * 100
    )
})

missing_before = missing_before[
    missing_before["Missing_Count"] > 0
].sort_values(
    by="Missing_Count",
    ascending=False
)

display(missing_before)
```

```python
# ============================================================
# 3.5 IDENTIFY LITERAL 'NA' VALUES
# ============================================================

print("\n" + "=" * 70)
print("LITERAL 'NA' VALUES")
print("=" * 70)

literal_na_before = (
    df.astype(str)
    .apply(
        lambda column: column.str.strip().str.lower().eq("na")
    )
    .sum()
)

literal_na_before = (
    literal_na_before[
        literal_na_before > 0
    ]
    .sort_values(ascending=False)
    .to_frame("Literal_NA_Count")
)

display(literal_na_before)
```

```python
# ============================================================
# 3.6 STANDARDIZE MISSING VALUE REPRESENTATIONS
# ============================================================

# Convert common representations of missing data into NaN.

missing_values = [
    "na",
    "NA",
    "N/A",
    "n/a",
    "null",
    "NULL",
    "None",
    "none",
    ""
]

df = df.replace(
    missing_values,
    np.nan
)
```

```python
# ============================================================
# 3.7 CHECK MISSING VALUES AFTER STANDARDIZATION
# ============================================================

print("\n" + "=" * 70)
print("MISSING VALUES AFTER STANDARDIZATION")
print("=" * 70)

missing_after_standardization = pd.DataFrame({
    "Column": df.columns,
    "Missing_Count": df.isnull().sum().values,
    "Missing_Percentage": (
        df.isnull().mean().values * 100
    )
})

missing_after_standardization = (
    missing_after_standardization[
        missing_after_standardization["Missing_Count"] > 0
    ]
    .sort_values(
        by="Missing_Count",
        ascending=False
    )
)

display(missing_after_standardization)
```

```python

# ============================================================
# 3.8 CHECK DUPLICATES
# ============================================================

print("\n" + "=" * 70)
print("DUPLICATE RECORDS")
print("=" * 70)

duplicate_count_before = df.duplicated().sum()

print(
    f"Duplicate rows found: "
    f"{duplicate_count_before:,}"
)
```

```python
# ============================================================
# 3.9 DISPLAY DUPLICATE RECORDS
# ============================================================

if duplicate_count_before > 0:

    print("\nSample duplicate records:")

    display(
        df[df.duplicated(keep=False)]
        .sort_values(by=df.columns.tolist())
        .head(20)
    )

else:

    print("No duplicate records found.")
```

```python
# ============================================================
# 3.10 REMOVE EXACT DUPLICATES
# ============================================================

df = df.drop_duplicates().reset_index(drop=True)

print("\nDuplicate records removed.")
```

```python
# ============================================================
# 3.11 CHECK DATA TYPES
# ============================================================

print("\n" + "=" * 70)
print("DATA TYPES BEFORE CONVERSION")
print("=" * 70)

display(
   df.dtypes
)

```

```python
# ============================================================
# 3.12 CONVERT NUMERICAL COLUMNS
# ============================================================

# These variables represent numerical quantities.
# errors='coerce' converts invalid entries into NaN,
# allowing us to identify them later.

numerical_columns_expected = [
    "Number_of_vehicles_involved",
    "Number_of_casualties"
]

for column in numerical_columns_expected:

    if column in df.columns:

        df[column] = pd.to_numeric(
            df[column],
            errors="coerce"
        )
```

```python
# ============================================================
# 3.13 CONVERT TIME COLUMN
# ============================================================

print("\n" + "=" * 70)
print("TIME COLUMN CONVERSION")
print("=" * 70)

if "Time" in df.columns:

    print("Time data type before conversion:")
    print(df["Time"].dtype)

    df["Time"] = pd.to_datetime(
        df["Time"],
        format="%H:%M:%S",
        errors="coerce"
    )

    print("\nTime data type after conversion:")
    print(df["Time"].dtype)
```

```python
# ============================================================
# 3.14 CHECK INVALID TIME VALUES
# ============================================================

if "Time" in df.columns:

    invalid_time_count = df["Time"].isna().sum()

    print(
        f"\nInvalid / missing time values: "
        f"{invalid_time_count:,}"
    )
```

```python
# ============================================================
# 3.15 STANDARDIZE TEXT COLUMNS
# ============================================================

# Remove unnecessary spaces from categorical/text variables.

text_columns = df.select_dtypes(
    include="object"
).columns

for column in text_columns:

    df[column] = (
        df[column]
        .str.strip()
    )
```

```python

# ============================================================
# 3.16 CHECK UNIQUE VALUES OF IMPORTANT CATEGORICAL COLUMNS
# ============================================================

important_categorical_columns = [
    "Day_of_week",
    "Age_band_of_driver",
    "Sex_of_driver",
    "Educational_level",
    "Driving_experience",
    "Type_of_vehicle",
    "Vehicle_driver_relation",
    "Area_accident_occured",
    "Lanes_or_Medians",
    "Road_allignment",
    "Types_of_Junction",
    "Road_surface_type",
    "Road_surface_conditions",
    "Light_conditions",
    "Weather_conditions",
    "Vehicle_movement",
    "Accident_severity"
]

print("\n" + "=" * 70)
print("IMPORTANT CATEGORICAL VARIABLES")
print("=" * 70)

for column in important_categorical_columns:

    if column in df.columns:

        print(f"\n{column}")
        print("-" * 50)

        display(
            df[column]
            .value_counts(dropna=False)
            .to_frame("Count")
        )
```

```python
# ============================================================
# 3.17 CHECK NUMERICAL VARIABLES FOR INVALID VALUES
# ============================================================

print("\n" + "=" * 70)
print("NUMERICAL VARIABLE VALIDATION")
print("=" * 70)

# Number of vehicles should be positive
if "Number_of_vehicles_involved" in df.columns:

    print("\nNumber of vehicles involved:")

    print(
        "Values <= 0:",
        (df["Number_of_vehicles_involved"] <= 0).sum()
    )


# Number of casualties should be positive
if "Number_of_casualties" in df.columns:

    print("\nNumber of casualties:")

    print(
        "Values <= 0:",
        (df["Number_of_casualties"] <= 0).sum()
    )
```

```python
# ============================================================
# 3.18 CHECK ACCIDENT SEVERITY VALUES
# ============================================================

print("\n" + "=" * 70)
print("ACCIDENT SEVERITY VALIDATION")
print("=" * 70)

display(
    df["Accident_severity"]
    .value_counts(dropna=False)
)
```

```python
# ============================================================
# 3.19 MISSING VALUE TREATMENT
# ============================================================

print("\n" + "=" * 70)
print("MISSING VALUE TREATMENT")
print("=" * 70)

# Display missing values before treatment

missing_summary = pd.DataFrame({
    "Column": df.columns,
    "Missing_Count": df.isnull().sum().values,
    "Missing_Percentage": (
        df.isnull().mean().values * 100
    )
})

missing_summary = (
    missing_summary[
        missing_summary["Missing_Count"] > 0
    ]
    .sort_values(
        by="Missing_Count",
        ascending=False
    )
)

display(missing_summary)


# ============================================================
# IMPORTANT:
# DO NOT AUTOMATICALLY DROP OR IMPUTE ALL MISSING VALUES YET.
# ============================================================

print(
    "\nMissing values identified."
)

print(
    "The appropriate treatment will depend on the variable "
    "and its analytical importance."
)

```

```python

# ============================================================
# 3.20 CREATE CLEANED DATASET STATUS
# ============================================================

print("\n" + "=" * 70)
print("CURRENT CLEANED DATASET STATUS")
print("=" * 70)

print(f"Rows    : {df.shape[0]:,}")
print(f"Columns : {df.shape[1]:,}")

print(
    f"\nRemaining duplicate rows: "
    f"{df.duplicated().sum():,}"
)
```

```python
# ============================================================
# 3.21 SAVE CLEANED DATASET
# ============================================================

# We will save the current cleaned version.
# Further feature engineering will happen later.

df_cleaned = df.copy()


# ============================================================
# 3.22 FINAL CLEANING SUMMARY
# ============================================================

print("\n" + "=" * 70)
print("STEP 3 - INITIAL DATA CLEANING COMPLETED")
print("=" * 70)

print(
    f"Original rows : {df_original.shape[0]:,}"
)

print(
    f"Current rows  : {df.shape[0]:,}"
)

print(
    f"Rows removed  : "
    f"{df_original.shape[0] - df.shape[0]:,}"
)

print(
    f"Columns       : {df.shape[1]:,}"
)

print("\nData has been standardized and validated.")
print("Remaining missing values will be treated based on column importance.")
```
